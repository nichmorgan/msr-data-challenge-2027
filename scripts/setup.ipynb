Import the dataset from huggingface, datasets should be cached locally so importing them will be much faster after the first time.

Default cache locations:
- Linux/macOS: ~/.cache/huggingface/datasets
- Windows: C:\Users\<username>\.cache\huggingface\datasets

You can manually set the cache dir in the load_dataset function

```python
from datasets import load_dataset
```

```python
artifacts = load_dataset("mvaccargiu/gitskills", "artifacts", split="train")

print(artifacts[0])
```

```python
siblings = load_dataset("mvaccargiu/gitskills", "artifact_siblings", split="train")

print(siblings[0])
```

```python
repos = load_dataset("mvaccargiu/gitskills", "repos", split="train")

print(repos[0])
```

```python
runs = load_dataset("mvaccargiu/gitskills", "mining_runs", split="train")

print(runs[0])
```
