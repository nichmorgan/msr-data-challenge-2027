```python
import polars as pl

from tool_extraction import artifacts, artifact_siblings, repos, mining_runs
```

## License analysis

```python
licenses = (
    repos.select(
        pl.col("license")
        .str.strip_chars()
        .str.to_lowercase()
        .fill_null("unlicense")
        .replace("", "unlicense")
        .alias("license")
    )
    .unique()
    .sort("license")
    .head(50)
    .collect()
)

with pl.Config(tbl_rows=-1):
    display(licenses)
```
